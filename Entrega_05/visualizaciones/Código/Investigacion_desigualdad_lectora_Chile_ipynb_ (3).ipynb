```python
# Seleccionar estudiantes con quintil socioeconómico disponible
pisa_quintiles = pisa[
    pisa["quintil_escs_chile"] != "Sin dato"
].copy()

# Construir tabla dinámica de puntajes
tabla_puntajes = pisa_quintiles.pivot(
    index=["quintil_escs_chile", "sexo"],
    columns="anio_pisa",
    values="puntaje_lectura_promedio"
)

# Calcular variaciones entre ciclos
tabla_puntajes["Variacion 2012-2015"] = (
    tabla_puntajes[2015] - tabla_puntajes[2012]
)

tabla_puntajes["Variacion 2015-2018"] = (
    tabla_puntajes[2018] - tabla_puntajes[2015]
)

tabla_puntajes["Variacion 2018-2022"] = (
    tabla_puntajes[2022] - tabla_puntajes[2018]
)

tabla_puntajes["Variacion 2012-2022"] = (
    tabla_puntajes[2022] - tabla_puntajes[2012]
)

# Mostrar resultados
print("EVOLUCIÓN DEL RENDIMIENTO LECTOR")
display(tabla_puntajes.round(2))
```

```python
import numpy as np

# Organizar los errores estándar por año, sexo y quintil
tabla_errores = pisa_quintiles.pivot(
    index=["quintil_escs_chile", "sexo"],
    columns="anio_pisa",
    values="error_estandar_puntaje"
)

# Crear una tabla para comparar los cambios observados
analisis_cambios = pd.DataFrame()

analisis_cambios["Puntaje 2018"] = tabla_puntajes[2018]
analisis_cambios["Puntaje 2022"] = tabla_puntajes[2022]

# Calcular la variación de puntajes entre ambos años
analisis_cambios["Variacion"] = (
    tabla_puntajes[2022] - tabla_puntajes[2018]
)

# Identificar la dirección del cambio observado
analisis_cambios["Cambio observado"] = np.select(
    [
        analisis_cambios["Variacion"] > 0,
        analisis_cambios["Variacion"] < 0,
        analisis_cambios["Variacion"] == 0
    ],
    [
        "Aumento del puntaje",
        "Disminucion del puntaje",
        "Sin cambio"
    ],
    default="Sin datos"
)

# Los cambios se interpretan de manera descriptiva.
# No se incorpora el error de enlace entre ciclos PISA.

print("ANALISIS DESCRIPTIVO DE LOS CAMBIOS ENTRE 2018 Y 2022")

display(analisis_cambios.round(2))
```

```python
# Crear una copia para trabajar sin modificar la base original
lectura = enpccl.copy()

# Agrupar las frecuencias de lectura
categorias_lectura = {
    "Al menos una vez al día": "Lectura frecuente",
    "Al menos una vez a la semana": "Lectura frecuente",
    "Al menos una vez al mes": "Lectura ocasional",
    "Al menos una vez en 3 meses": "Lectura ocasional",
    "Al menos una vez en 6 meses": "Lectura ocasional",
    "Al menos una vez al año": "Lectura ocasional",
    "No leo": "No lee libros"
}

lectura["habito_lector"] = (
    lectura["frecuencia_libros"].map(categorias_lectura)
)

# Conservar respuestas válidas
lectura_valida = lectura.dropna(
    subset=["habito_lector"]
).copy()

# Comprobar los factores de expansión
peso = "factor_expansion_persona"

lectura_valida[peso] = pd.to_numeric(
    lectura_valida[peso], errors="coerce"
)

assert np.isfinite(lectura_valida[peso]).all()
assert (lectura_valida[peso] > 0).all()

print("Registros originales:", len(lectura))
print("Registros validos:", len(lectura_valida))

# Función para calcular porcentajes ponderados
def tabla_ponderada(grupos):

    tabla = lectura_valida.pivot_table(
        index=grupos,
        columns="habito_lector",
        values=peso,
        aggfunc="sum",
        fill_value=0
    )

    porcentajes = tabla.div(
        tabla.sum(axis=1), axis=0
    ) * 100

    return porcentajes[
        ["Lectura frecuente",
         "Lectura ocasional",
         "No lee libros"]
    ].round(2)

# Cruce 1: sexo y hábitos lectores
tabla_sexo = tabla_ponderada(["sexo"])

print("\nHABITOS LECTORES SEGUN SEXO (%)")
display(tabla_sexo)

# Cruce 2: nivel socioeconómico y hábitos lectores
tabla_nse = tabla_ponderada(
    ["nivel_socioeconomico"]
)

print("\nHABITOS LECTORES SEGUN NSE (%)")
display(tabla_nse)

# Cruce 3: sexo, NSE y hábitos lectores
tabla_sexo_nse = tabla_ponderada(
    ["nivel_socioeconomico", "sexo"]
)

print("\nHABITOS LECTORES SEGUN NSE Y SEXO (%)")
display(tabla_sexo_nse)
```

```python
# Cruce de edad y hábitos lectores
tabla_edad = tabla_ponderada(
    ["grupo_edad"]
)

print("HABITOS LECTORES SEGUN EDAD (%)")
display(tabla_edad)

# Cruce de edad, sexo y hábitos lectores
tabla_edad_sexo = tabla_ponderada(
    ["grupo_edad", "sexo"]
)

print("\nHABITOS LECTORES SEGUN EDAD Y SEXO (%)")
display(tabla_edad_sexo)

# Seleccionar solamente a los jóvenes
jovenes = lectura_valida[
    lectura_valida["grupo_edad"] ==
    "15 a 29 (Jóvenes)"
].copy()

print("\nCantidad de jóvenes con respuesta válida:")
print(len(jovenes))

# Cruzar sexo y NSE entre los jóvenes
tabla_jovenes = jovenes.pivot_table(
    index=["nivel_socioeconomico", "sexo"],
    columns="habito_lector",
    values="factor_expansion_persona",
    aggfunc="sum",
    fill_value=0
)

tabla_jovenes = tabla_jovenes.div(
    tabla_jovenes.sum(axis=1), axis=0
) * 100

print("\nHABITOS LECTORES DE LOS JOVENES (%)")
display(tabla_jovenes.round(2))

# Revisar tamaño de muestra por segmento
print("\nCANTIDAD DE JOVENES POR SEGMENTO")
display(
    jovenes.groupby(
        ["nivel_socioeconomico", "sexo"]
    ).size()
)
```

```python
# PASO 6.8: AUTOPERCEPCION LECTORA

import pandas as pd

# Crear una copia de la base
percepcion = enpccl.copy()

# Identificar las dimensiones lectoras
dimensiones = {
    "acuerdo_ritmo_lectora": "Dificultad de ritmo",
    "acuerdo_comprension_lectora": "Dificultad de comprension",
    "acuerdo_concentracion_lectora": "Dificultad de concentracion",
    "acuerdo_paciencia_lectora": "Falta de paciencia",
    "acuerdo_vision_lectora": "Dificultad de vision"
}

# Categorias de respuesta
respuestas = {
    "Muy de acuerdo": 0,
    "De acuerdo": 0,
    "En desacuerdo": 1,
    "Muy en desacuerdo": 1
}

# Crear indicadores de dificultad
for variable, nombre in dimensiones.items():
    percepcion[nombre] = percepcion[variable].map(respuestas)

# Factor de expansion
peso = "factor_expansion_persona"

percepcion[peso] = pd.to_numeric(
    percepcion[peso], errors="coerce"
)

# Funcion para calcular porcentajes ponderados
def porcentaje_dificultad(datos):

    resultados = {}

    for nombre in dimensiones.values():

        validos = datos.dropna(subset=[nombre, peso])
        validos = validos[validos[peso] > 0]

        denominador = validos[peso].sum()

        if denominador > 0:
            resultados[nombre] = (
                (validos[nombre] * validos[peso]).sum()
                / denominador * 100
            )
        else:
            resultados[nombre] = float("nan")

    return pd.Series(resultados)

# Resultados generales
tabla_percepcion_general = porcentaje_dificultad(
    percepcion
)

print("DIFICULTADES LECTORAS GENERALES (%)")
display(tabla_percepcion_general.round(2))

# Resultados segun sexo
tabla_percepcion_sexo = pd.DataFrame({
    sexo: porcentaje_dificultad(grupo)
    for sexo, grupo in percepcion.groupby("sexo")
}).T

print("\nDIFICULTADES LECTORAS SEGUN SEXO (%)")
display(tabla_percepcion_sexo.round(2))

# Resultados segun edad
tabla_percepcion_edad = pd.DataFrame({
    edad: porcentaje_dificultad(grupo)
    for edad, grupo in percepcion.groupby("grupo_edad")
}).T

print("\nDIFICULTADES LECTORAS SEGUN EDAD (%)")
display(tabla_percepcion_edad.round(2))

# Resultados segun nivel socioeconomico
tabla_percepcion_nse = pd.DataFrame({
    nse: porcentaje_dificultad(grupo)
    for nse, grupo in percepcion.groupby(
        "nivel_socioeconomico"
    )
}).T

print("\nDIFICULTADES LECTORAS SEGUN NSE (%)")
display(tabla_percepcion_nse.round(2))

# Seleccionar jovenes de 15 a 29 años
jovenes_percepcion = percepcion[
    percepcion["grupo_edad"] == "15 a 29 (Jóvenes)"
].copy()

# Cruzar sexo y NSE entre los jovenes
tabla_percepcion_jovenes = pd.DataFrame({
    grupo: porcentaje_dificultad(datos)
    for grupo, datos in jovenes_percepcion.groupby(
        ["nivel_socioeconomico", "sexo"]
    )
}).T

tabla_percepcion_jovenes.index.names = [
    "Nivel socioeconomico", "Sexo"
]

print("\nDIFICULTADES LECTORAS DE LOS JOVENES (%)")
display(tabla_percepcion_jovenes.round(2))
```

```python
# PASO 6.9: MOTIVACIONES LECTORAS

import pandas as pd

motivaciones = enpccl.copy()

peso = "factor_expansion_persona"
variable = "motivo_principal_lectura"

# Convertir el factor de expansion a numero
motivaciones[peso] = pd.to_numeric(
    motivaciones[peso], errors="coerce"
)

# Excluir respuestas no validas
motivaciones = motivaciones[
    ~motivaciones[variable].isin([
        "No aplica",
        "No sabe",
        "No responde",
        "Sin dato"
    ])
].copy()

motivaciones = motivaciones.dropna(
    subset=[variable, peso]
)

motivaciones = motivaciones[
    motivaciones[peso] > 0
]

# Funcion para calcular porcentajes ponderados
def tabla_motivaciones(datos, grupos):

    tabla = datos.groupby(
        grupos + [variable]
    )[peso].sum().unstack(fill_value=0)

    return tabla.div(
        tabla.sum(axis=1), axis=0
    ) * 100

# 1. Motivaciones generales
tabla_motivos_general = (
    motivaciones.groupby(variable)[peso].sum()
    / motivaciones[peso].sum() * 100
).sort_values(ascending=False)

print("MOTIVACIONES LECTORAS GENERALES (%)")
display(tabla_motivos_general.round(2))

# 2. Motivaciones segun sexo
tabla_motivos_sexo = tabla_motivaciones(
    motivaciones, ["sexo"]
)

print("\nMOTIVACIONES SEGUN SEXO (%)")
display(tabla_motivos_sexo.round(2))

# 3. Motivaciones segun NSE
tabla_motivos_nse = tabla_motivaciones(
    motivaciones, ["nivel_socioeconomico"]
)

print("\nMOTIVACIONES SEGUN NSE (%)")
display(tabla_motivos_nse.round(2))

# 4. Seleccionar jovenes
jovenes_motivaciones = motivaciones[
    motivaciones["grupo_edad"] ==
    "15 a 29 (Jóvenes)"
].copy()

# Motivaciones de los jovenes segun sexo
tabla_motivos_jovenes_sexo = tabla_motivaciones(
    jovenes_motivaciones, ["sexo"]
)

print("\nMOTIVACIONES DE JOVENES SEGUN SEXO (%)")
display(tabla_motivos_jovenes_sexo.round(2))

# 5. Motivaciones de jovenes segun sexo y NSE
tabla_motivos_jovenes = tabla_motivaciones(
    jovenes_motivaciones,
    ["nivel_socioeconomico", "sexo"]
)

print("\nMOTIVACIONES DE JOVENES SEGUN NSE Y SEXO (%)")
display(tabla_motivos_jovenes.round(2))

# 6. Aislar el gusto por leer
tabla_gusto_lectura = tabla_motivos_jovenes[
    ["Porque me gusta"]
].copy()

print("\nJOVENES QUE LEEN PRINCIPALMENTE POR GUSTO (%)")
display(tabla_gusto_lectura.round(2))

# 7. Cantidad de respuestas validas por segmento
print("\nCANTIDAD DE JOVENES POR SEGMENTO")

display(
    jovenes_motivaciones.groupby(
        ["nivel_socioeconomico", "sexo"]
    ).size()
)
```

```python
# PASO 6.10: MOTIVOS DE NO LECTURA

import pandas as pd

no_lectura = enpccl.copy()

peso = "factor_expansion_persona"
variable_no = "motivo_principal_no_lectura"

# Convertir factor de expansion a numero
no_lectura[peso] = pd.to_numeric(
    no_lectura[peso], errors="coerce"
)

# Excluir respuestas no validas
no_lectura = no_lectura[
    ~no_lectura[variable_no].isin([
        "No aplica",
        "No sabe",
        "No responde",
        "Sin dato"
    ])
].copy()

no_lectura = no_lectura.dropna(
    subset=[variable_no, peso]
)

no_lectura = no_lectura[
    no_lectura[peso] > 0
]

# Funcion para calcular porcentajes ponderados
def tabla_no_lectura(datos, grupos):

    tabla = datos.groupby(
        grupos + [variable_no]
    )[peso].sum().unstack(fill_value=0)

    return tabla.div(
        tabla.sum(axis=1), axis=0
    ) * 100

# 1. Motivos generales
tabla_no_general = (
    no_lectura.groupby(variable_no)[peso].sum()
    / no_lectura[peso].sum() * 100
).sort_values(ascending=False)

print("MOTIVOS GENERALES DE NO LECTURA (%)")
display(tabla_no_general.round(2))

# 2. Motivos segun sexo
tabla_no_sexo = tabla_no_lectura(
    no_lectura, ["sexo"]
)

print("\nMOTIVOS DE NO LECTURA SEGUN SEXO (%)")
display(tabla_no_sexo.round(2))

# 3. Motivos segun nivel socioeconomico
tabla_no_nse = tabla_no_lectura(
    no_lectura, ["nivel_socioeconomico"]
)

print("\nMOTIVOS DE NO LECTURA SEGUN NSE (%)")
display(tabla_no_nse.round(2))

# 4. Motivos segun edad
tabla_no_edad = tabla_no_lectura(
    no_lectura, ["grupo_edad"]
)

print("\nMOTIVOS DE NO LECTURA SEGUN EDAD (%)")
display(tabla_no_edad.round(2))

# 5. Seleccionar jovenes
jovenes_no_lectura = no_lectura[
    no_lectura["grupo_edad"] ==
    "15 a 29 (Jóvenes)"
].copy()

print("\nCANTIDAD TOTAL DE RESPUESTAS VALIDAS")
print(len(no_lectura))

print("\nCANTIDAD DE JOVENES CON RESPUESTAS VALIDAS")
print(len(jovenes_no_lectura))

# 6. Motivos de no lectura entre jovenes
if len(jovenes_no_lectura) > 0:

    tabla_no_jovenes = (
        jovenes_no_lectura.groupby(variable_no)[peso].sum()
        / jovenes_no_lectura[peso].sum() * 100
    ).sort_values(ascending=False)

    print("\nMOTIVOS DE NO LECTURA DE JOVENES (%)")
    display(tabla_no_jovenes.round(2))

    print("\nCANTIDAD DE JOVENES SEGUN NSE Y SEXO")

    display(
        jovenes_no_lectura.groupby(
            ["nivel_socioeconomico", "sexo"]
        ).size()
    )
```

```python
# PASO 6.11: REDES SOCIALES Y DIFICULTADES LECTORAS

import pandas as pd

redes = enpccl.copy()

peso = "factor_expansion_persona"
variable_redes = "redes_sociales_mientras_lee"

# Preparar factor de expansion
redes[peso] = pd.to_numeric(
    redes[peso], errors="coerce"
)

# Seleccionar respuestas validas
redes = redes[
    redes[variable_redes].isin(["Sí", "No"])
].copy()

redes = redes.dropna(subset=[peso])
redes = redes[redes[peso] > 0]

# Crear indicador de uso de redes sociales
redes["usa_redes"] = (
    redes[variable_redes] == "Sí"
).astype(int)

# Funcion para calcular porcentajes ponderados
def porcentaje_ponderado(datos, variable):

    return (
        (datos[variable] * datos[peso]).sum()
        / datos[peso].sum()
    ) * 100

# 1. Uso general de redes sociales
print("USO DE REDES SOCIALES MIENTRAS SE LEE (%)")

print(round(
    porcentaje_ponderado(redes, "usa_redes"), 2
))

# 2. Uso segun edad
print("\nUSO DE REDES SOCIALES SEGUN EDAD (%)")

display(
    redes.groupby("grupo_edad").apply(
        lambda x: porcentaje_ponderado(x, "usa_redes")
    ).round(2)
)

# 3. Seleccionar jovenes
jovenes_redes = redes[
    redes["grupo_edad"] == "15 a 29 (Jóvenes)"
].copy()

print("\nCANTIDAD DE JOVENES CON RESPUESTA VALIDA")
print(len(jovenes_redes))

# 4. Uso de redes segun sexo
print("\nUSO DE REDES EN JOVENES SEGUN SEXO (%)")

display(
    jovenes_redes.groupby("sexo").apply(
        lambda x: porcentaje_ponderado(x, "usa_redes")
    ).round(2)
)

# 5. Uso segun NSE y sexo
print("\nUSO DE REDES EN JOVENES SEGUN NSE Y SEXO (%)")

display(
    jovenes_redes.groupby(
        ["nivel_socioeconomico", "sexo"]
    ).apply(
        lambda x: porcentaje_ponderado(x, "usa_redes")
    ).round(2)
)

# 6. Variables de dificultades lectoras
variables_dificultad = {
    "acuerdo_concentracion_lectora":
        "Dificultad de concentracion",

    "acuerdo_comprension_lectora":
        "Dificultad de comprension",

    "acuerdo_paciencia_lectora":
        "Falta de paciencia",

    "acuerdo_ritmo_lectora":
        "Dificultad de ritmo"
}

resultados = []

respuestas_validas = [
    "De acuerdo",
    "Muy de acuerdo",
    "En desacuerdo",
    "Muy en desacuerdo"
]

# 7. Cruzar redes sociales con dificultades
for variable, nombre in variables_dificultad.items():

    for respuesta in ["Sí", "No"]:

        grupo = jovenes_redes[
            (jovenes_redes[variable_redes] == respuesta)
            & jovenes_redes[variable].isin(
                respuestas_validas
            )
        ].copy()

        if len(grupo) == 0:
            continue

        grupo["dificultad"] = grupo[variable].isin([
            "En desacuerdo",
            "Muy en desacuerdo"
        ]).astype(int)

        porcentaje = porcentaje_ponderado(
            grupo, "dificultad"
        )

        resultados.append({
            "Dificultad": nombre,
            "Usa redes mientras lee": respuesta,
            "Porcentaje": porcentaje,
            "Cantidad de encuestados": len(grupo)
        })

# 8. Crear tabla de resultados
tabla_redes_dificultades = pd.DataFrame(resultados)

print("\nREDES SOCIALES Y DIFICULTADES LECTORAS EN JOVENES")

display(
    tabla_redes_dificultades.round(2)
)

# 9. Comparar ambos grupos
comparacion_redes = tabla_redes_dificultades.pivot(
    index="Dificultad",
    columns="Usa redes mientras lee",
    values="Porcentaje"
)

comparacion_redes["Brecha en puntos porcentuales"] = (
    comparacion_redes["Sí"]
    - comparacion_redes["No"]
)

print("\nCOMPARACION DE DIFICULTADES LECTORAS (%)")

display(comparacion_redes.round(2))

# 10. Revisar tamaño de los grupos
print("\nCANTIDAD DE JOVENES SEGUN USO DE REDES")

display(
    jovenes_redes[variable_redes].value_counts()
)
```

### 6.12.3. Evolución de la lectura por ocio entre 2014 y 2024

Se analiza la evolución de la cantidad de libros leídos por gusto, entretención u ocio durante los últimos 12 meses, utilizando las bases armonizadas de la ECL 2014 y la ENPCCL 2024.

El análisis considera exclusivamente a los lectores de ocio incluidos en la base y utiliza los factores de expansión originales de cada encuesta.

Se calculan promedios ponderados, medianas ponderadas y distribuciones porcentuales por cantidad de libros leídos. La comparación es descriptiva y exploratoria, debido a las diferencias metodológicas entre ambas encuestas.

```python
import pandas as pd
import numpy as np

# Preparar una copia de la base
ocio_valido = ocio.copy()

# Seleccionar registros válidos
ocio_valido = ocio_valido[
    (ocio_valido["libros_ocio_12_meses"] > 0) &
    (ocio_valido["factor_expansion_persona"] > 0)
].copy()

# Función para calcular la mediana ponderada
def mediana_ponderada(grupo):
    ordenado = grupo.sort_values("libros_ocio_12_meses")
    pesos = ordenado["factor_expansion_persona"]
    acumulado = pesos.cumsum()
    limite = pesos.sum() / 2

    return ordenado.loc[
        acumulado >= limite,
        "libros_ocio_12_meses"
    ].iloc[0]

# Calcular resultados por año
resultados_ocio = []

for anio, grupo in ocio_valido.groupby("anio_encuesta"):

    promedio = np.average(
        grupo["libros_ocio_12_meses"],
        weights=grupo["factor_expansion_persona"]
    )

    resultados_ocio.append({
        "Año": anio,
        "Promedio ponderado de libros": promedio,
        "Mediana ponderada de libros": mediana_ponderada(grupo),
        "Cantidad de encuestados": len(grupo)
    })

tabla_ocio_anual = pd.DataFrame(resultados_ocio)

print("CANTIDAD DE LIBROS LEÍDOS POR OCIO")
display(tabla_ocio_anual.round(2))

# Distribución ponderada por cantidad de libros
tabla_distribucion_ocio = ocio_valido.pivot_table(
    index="tramo_libros_ocio",
    columns="anio_encuesta",
    values="factor_expansion_persona",
    aggfunc="sum",
    fill_value=0
)

tabla_distribucion_ocio = (
    tabla_distribucion_ocio /
    tabla_distribucion_ocio.sum()
) * 100

orden_tramos = [
    "1 a 2 libros",
    "3 a 5 libros",
    "6 a 10 libros",
    "11 a 20 libros",
    "21 libros o más"
]

tabla_distribucion_ocio = tabla_distribucion_ocio.reindex(
    orden_tramos
)

tabla_distribucion_ocio["Cambio en puntos porcentuales"] = (
    tabla_distribucion_ocio[2024] -
    tabla_distribucion_ocio[2014]
)

print("DISTRIBUCIÓN POR CANTIDAD DE LIBROS (%)")
display(tabla_distribucion_ocio.round(2))
```

Se realiza una prueba de sensibilidad limitando temporalmente las cantidades a 24 y 50 libros anuales, sin modificar la base de datos original.

El objetivo es comprobar cuánto dependen los resultados de los valores altos y determinar qué indicadores son más adecuados para comunicar los hallazgos de la investigación.

```python
# Revisión de valores extremos y sensibilidad

resultados_sensibilidad = []

for (anio, sexo), grupo in ocio_jovenes.groupby(
    ["anio_encuesta", "sexo"]
):

    libros = grupo["libros_ocio_12_meses"]
    pesos = grupo["factor_expansion_persona"]

    resultados_sensibilidad.append({
        "Año": anio,
        "Sexo": sexo,
        "Encuestados": len(grupo),
        "Máximo de libros": libros.max(),
        "Casos sobre 24 libros": (libros > 24).sum(),
        "Casos sobre 50 libros": (libros > 50).sum(),
        "Promedio original": np.average(
            libros, weights=pesos
        ),
        "Promedio con límite de 24": np.average(
            libros.clip(upper=24), weights=pesos
        ),
        "Promedio con límite de 50": np.average(
            libros.clip(upper=50), weights=pesos
        ),
        "Mediana ponderada": mediana_ponderada(grupo)
    })

tabla_sensibilidad = pd.DataFrame(
    resultados_sensibilidad
)

print("REVISIÓN DE SENSIBILIDAD")
display(tabla_sensibilidad.round(2))


# Revisar las diez cantidades más altas de cada grupo

print("VALORES MÁS ALTOS POR AÑO Y SEXO")

for (anio, sexo), grupo in ocio_jovenes.groupby(
    ["anio_encuesta", "sexo"]
):

    print(f"\nAño: {anio} | Sexo: {sexo}")

    display(
        grupo.nlargest(
            10, "libros_ocio_12_meses"
        )[
            [
                "libros_ocio_12_meses",
                "factor_expansion_persona"
            ]
        ]
    )
```

Estos resultados sugieren una mayor concentración de lectoras en los extremos de la distribución, aunque se requiere un análisis estadístico adicional para determinar la significancia de estos cambios.

Estos resultados complementan el análisis de las desigualdades en comprensión lectora realizado mediante las bases de datos PISA.

Los hábitos lectores y el desempeño en comprensión lectora corresponden a dimensiones diferentes. Por lo tanto, estas encuestas permiten aportar contexto a la investigación, pero no establecer una relación causal entre la cantidad de libros leídos y los resultados PISA.
